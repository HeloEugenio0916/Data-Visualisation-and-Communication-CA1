#### Selecting occupation codes
###### '15-1221'	Computer and Information Research Scientists	
###### '15-2051'	Data Scientists	
###### '15-1252'	Software Developers	

```python
occupations = ["15-1221", "15-2051", "15-1252"]
```

###### The dataset contains 1,138 different occupation titles, so it was important to narrow down the data and focus only on the required ones, such as (AI, Data Science and Software Development).
###### Using their own occupation codes made it easier to select, analyse, and compare. 

#### Searching for a specific occupation code

```python
dataset.loc[dataset["OCC_CODE"] == "15-1252", ["OCC_CODE", "OCC_TITLE"]]
```

#### Filtering the Dataset by Occupation (splitting into smaller parts) 

```python
tech = dataset.loc[dataset["OCC_CODE"].isin(occupations)]
tech.shape
```

###### A smaller dataset called tech was created to reduce the number of records from 413,327 to 1,768, keeping only the three selected occupations, making it easier to analyse their employment and salaries. 

#### Exploring Geographical Records

```python
tech["AREA_TITLE"].value_counts()
```

#### Selecting National Employment Data
Instead of checking individual cities, I selected national employment across the entire USA. 

```python
national = tech.loc[(tech["AREA_TITLE"] == "U.S.") &
                    (tech["NAICS"] == "000000")]
```
