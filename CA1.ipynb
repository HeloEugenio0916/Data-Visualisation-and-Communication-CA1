#### Filtering the Dataset by Occupation (splitting into smaller parts) 

```python
tech = dataset.loc[dataset["OCC_CODE"].isin(occupations)]
tech.shape
```

###### A smaller dataset called tech was created to reduce the number of records from 413,327 to 1,768, keeping only the three selected occupations, making it easier to analyse their employment and salaries. 

#### Exploring Geographical Records

```python
tech["AREA_TITLE"].value_counts()
```

#### Selecting National Employment Data
Instead of checking individual cities, I selected national employment across the entire USA. 

```python
national = tech.loc[(tech["AREA_TITLE"] == "U.S.") &
                    (tech["NAICS"] == "000000")]
```

#### Checking National Employment Records

```python
national[["OCC_CODE", "OCC_TITLE", "TOT_EMP", "OWN_CODE"]]
```

### 1st Bar Chart = Comparing employment through main occupations
##### Objective and Chart choice
The main objective of this chart is to compare the total employment of the three required occupations: AI, Data Science and Software Development across the USA in 2023. 

The horizontal bar chart was chosen because it makes it easier to compare the employment numbers and read the occupation names clearly. 


```python
plt.figure(figsize=(8, 5))

colours = ["lightblue", "pink", "khaki"]

sns.barplot(x='TOT_EMP', y='OCC_TITLE', data=national,
            hue='OCC_TITLE', palette=colors, legend=False)

plt.title('Employment of Selected Technology Occupations in the USA (2023)', fontsize=11)

plt.xlabel('Total Employment', fontsize=12)
plt.ylabel('Occupation', fontsize=12)

plt.tight_layout()

plt.show()
```

##### Results and Analysis
According to the results, Software Developers had the highest employment, with approximately 1.66 million, followed by Data Scientists with 192.710 and Computer and Information Research scientists with 35,210.
