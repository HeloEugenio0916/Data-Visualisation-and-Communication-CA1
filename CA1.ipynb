### Histogram Chart = Annual Median Wage Distribution

##### Objective and Chart Choice

The objective of this visualisation was to check and compare the distribution of annual median wages for the required occupations across USA metropolitan areas in 2023.

The histogram chart was chosen because it groups numerical salary values into ranges. Making it easier to visualise where wages are more concentrated and how their distributions change between occupations. 

```python
metro = metro.copy()
metro["A_MEDIAN"] = pd.to_numeric(metro["A_MEDIAN"], errors="coerce")

metro["A_MEDIAN"].isnull().sum()

salary = metro.dropna(subset=["A_MEDIAN"])

salary.shape

plt.figure(figsize=(7, 5))

sns.histplot(
    data=salary,
    x="A_MEDIAN",
    hue="OCC_TITLE",
    bins=20,
    palette=["lightblue", "pink", "khaki"],
    alpha=0.8
)

plt.title("Distribution of Annual Median Wages in U.S. Metropolitan Areas (2023)")
plt.xlabel("Annual Median Wage (USD)")
plt.ylabel("Number of Metropolitan-Area Records")

plt.tight_layout()
plt.show()
```

##### Results and Analysis
    * Software Dev - Most salary records are concentrated between approximately $95,000 and $130,000
    * Data Scientist - between $75,000 and $105,000
    * Computer and Inf - Salaries that are more spread out, with $200,000

According to the histogram, Software Dev and Data Scientists have most of their annual median wages within smaller ranges. Meanwhile, Computer and Information Research Scientists show a wider distribution of wages across metropolitan areas. 

Overall, Research Scientists' annual median wages vary more between locations, while Software Developers and Data Scientists have more similar wage ranges.

### Chart 4: Software Developer Employment Share by U.S. State (2023)

The software developers were selected to investigate the geographical distribution across the USA in 2023, as they had the highest national employment among the three occupations. 


#### Checking all the different geographical areas available 

```python
dataset[["AREA_TYPE", "AREA_TITLE"]].drop_duplicates().head(20)
```

#### Filtering Software Developers by occupation code 

```python
# Filtering the dataset to include only USA state-level records and Software Dev
state_dev = dataset.loc[
    (dataset["AREA_TYPE"] == 2) &
    (dataset["OCC_CODE"] == "15-1252")
]
```

```python
# Displaying the first 10 records to check the occupation, industry, ownership and employment
state_dev[["AREA_TITLE", "OCC_TITLE", "NAICS", "OWN_CODE", "TOT_EMP"]].head(10)
```

#### Checking the number of State Records

51 unique geographical areas 
* 50 USA states // Washington, DC

```python
print(state_dev.shape)
print(state_dev["AREA_TITLE"].nunique())
```

#### Identifying the Top 5 States

The state-level employment records were sorted from highest to lowest to identify the five USA states with the largest number of Software Developers in 2023.

```python
top5 = state_dev.sort_values("TOT_EMP", ascending=False).head(5)

top5[["AREA_TITLE", "TOT_EMP"]]
```

#### Calculating total employment 

The employment estimates from all 51 areas were combined to calculate the total Software Developer employment in the USA in 2023. This total will be used to calculate each state's employment share

```python
total_emp = pd.to_numeric(state_dev["TOT_EMP"], errors="coerce").sum()

#total number of Software Developers
print(total_emp)
```

#### Calculating employment percentages
The percentage of Software Developers working in each of the top five states was calculated using the total employment in the USA. 

```python
top5["SHARE"] = pd.to_numeric(top5["TOT_EMP"]) / total_emp * 100

top5[["AREA_TITLE", "TOT_EMP", "SHARE"]]
```

#### Calculating other states left 
The remaining employment percentage was calculated by subtracting the top five states' combined percentage from 100%. This represents the other 46 areas.

```python
other_share = 100 - top5["SHARE"].sum()

print(other_share)
```

#### Preparing Data for the Pie Chart

```python
pie_data = pd.concat([
    top5[["AREA_TITLE", "SHARE"]],
    pd.DataFrame({
        "AREA_TITLE": ["Other States"],
        "SHARE": [other_share]
    })
], ignore_index=True)

pie_data
```

### Pie Chart = Software Developer Employment Share
##### Objective and Chart Choice
The objective was to identify which states in the USA had the highest percentage of software developers in 2023. 

A pie chart was chosen because it was the easiest to understand and better for visualising the distribution of employment across the top five states and the remaining areas.

```python
colors = ["#A8DADC", "#F4A7B9", "#F6D98F", "#B8A9E8", "#A9D6B5", "#D3D3D3"]

plt.figure(figsize=(7, 5))

plt.pie(
    pie_data["SHARE"],
    labels=pie_data["AREA_TITLE"],
    autopct="%1.1f%%",
    startangle=90,
    colors=colors
)

plt.title("Software Developer Employment Share by USA State (2023)")
plt.tight_layout()
plt.show()
```

#### Results and Analysis

According to the  pie chart, California has the highest employment percentage (18.4%), followed by Texas (8.4%) and New York (6.4%) 

This information is important because it shows which states employ the largest number of Software Developers and can help to have a better understanding of where job opportunities may be more concentrated. 

The remaining 56.3% represents the other 46 areas combined and is also widely distributed across the whole USA. 

### Chart 5 - Employment vs Annual Median Wage (2023)

##### Investigating the Relationship Between Employment and Annual Median Wages
This chart will check whether metropolitan areas with higher employment numbers also offer higher annual median wages across three technology occupations in the USA.

#### Preparing employment and salary data

```python
# Metropolitan area records for Software Developers
metro_dev = metro.loc[metro["OCC_CODE"] == "15-1252"].copy()
print(metro_dev.shape)

metro_dev[["AREA_TITLE", "OCC_TITLE", "TOT_EMP", "A_MEDIAN"]].head(10)
  
```

#### Preparing the data for comparison 

After removing any missing employment or wage values, 684 valid records remained to investigate whether areas with more employees may also have higher annual median wages. 

```python
# Create a copy of the metropolitan-area dataset
scatter_all = metro.copy()

# Making sure employment and wages are numbers
scatter_all["TOT_EMP"] = pd.to_numeric(scatter_all["TOT_EMP"], errors="coerce")
scatter_all["A_MEDIAN"] = pd.to_numeric(scatter_all["A_MEDIAN"], errors="coerce")

# Removing any missing values
scatter_all = scatter_all.dropna(subset=["TOT_EMP", "A_MEDIAN"])

# Check the number of records for each occupation
scatter_all["OCC_TITLE"].value_counts()
```

### Scatter Plot Chart = Exploring the Relationship Between Employment and Wages

For the final chart, the main objective was to investigate the relationship between employment numbers and annual median wages of the three occupations

A scatter plot was chosen because it helps to compare and understand two numerical values. Each dot represents a record, making it easier to identify the relationship and see whether higher employment is connected to higher wages. 


```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=scatter_all,
    x="TOT_EMP",
    y="A_MEDIAN",
    hue="OCC_TITLE",
    palette=["#5DADE2", "#F48FB1", "#F4D35E"],
    alpha=0.6
)

plt.title("Employment vs Annual Median Wage by Occupation (USA, 2023)")
plt.xlabel("Number of Employees")
plt.ylabel("Annual Median Wage (USD)")

plt.tight_layout()
plt.show()
```

##### Results and Analysis

According to the chart, Software Developers have the highest employment numbers in metropolitan areas, while Computer and information research scientists have some of the highest annual median wages. 

This suggest that areas with more employees do not always have higher annual median wages, and also helps us to understand that employment and wages can vary depending on the job occupation and location.
