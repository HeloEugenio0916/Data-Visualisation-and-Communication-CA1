### Histogram Chart = Annual Median Wage Distribution

##### Objective and Chart Choice

The objective of this visualisation was to check and compare the distribution of annual median wages for the required occupations across USA metropolitan areas in 2023.

The histogram chart was chosen because it groups numerical salary values into ranges. Making it easier to visualise where wages are more concentrated and how their distributions change between occupations. 

```python
metro = metro.copy()
metro["A_MEDIAN"] = pd.to_numeric(metro["A_MEDIAN"], errors="coerce")

metro["A_MEDIAN"].isnull().sum()

salary = metro.dropna(subset=["A_MEDIAN"])

salary.shape

plt.figure(figsize=(7, 5))

sns.histplot(
    data=salary,
    x="A_MEDIAN",
    hue="OCC_TITLE",
    bins=20,
    palette=["lightblue", "pink", "khaki"],
    alpha=0.8
)

plt.title("Distribution of Annual Median Wages in U.S. Metropolitan Areas (2023)")
plt.xlabel("Annual Median Wage (USD)")
plt.ylabel("Number of Metropolitan-Area Records")

plt.tight_layout()
plt.show()
```

##### Results and Analysis

    * Software Dev - Most salary records are concentrated between approximately $95,000 and $130,000
    * Data Scientist - between $75,000 and $105,000
    * Computer and Inf - Salaries that are more spread out, with $200,000

### Box Plot Chart = Annual Median Wage Comparison
##### Objective and Chart Choice

The objective of this visualisation is to compare and identify the annual median wages for the three roles chosen. 

A box plot was a great option to illustrate wage variation, compare median values and identify outliers. 

```python
plt.figure(figsize=(9, 4))

sns.boxplot(data=salary, x="A_MEDIAN", y="OCC_TITLE",
            hue="OCC_TITLE",
            palette=["lightblue", "pink", "khaki"],
            legend=False)

plt.title("Annual Median Wage Comparison in U.S. Metropolitan Areas (2023)")
plt.xlabel("Annual Median Wage (USD)")
plt.ylabel("Occupation")

plt.tight_layout()
plt.show()
```

##### Results and Analysis
The Box Plot chart shows us that Computer and Inf Scientist are the ones with the highest annual median wages, followed by Software Dev and Data Scientist. 

The chart also shows that Computer and Inf Scientist have greater wage variation, with an annual median wage of $233,110, and Data Scientist has the lowest typical wage estimates among the three occupations.

Annual median salary also varies by location, as metropolitan areas report different salary levels for the same occupation
